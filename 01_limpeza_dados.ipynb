```python
#dataframe
df = pd.read_excel('data/raw/dados_cesi.xlsx')
```

```python
#eliminando a fase experimental de 2023
df_cesi = df.set_index('data').sort_index()
df_cesi = df_cesi.loc['2004-01-01':]

df_cesi.head()
```

```python
df_etf = pd.read_excel('data/raw/yahoo_carteira_universo.xlsx', sheet_name='TLT')
df_etf['Date'] = pd.to_datetime(df_etf['Date'])
df_etf = df_etf.set_index('Date').sort_index()

df_alinhado = df_cesi.reindex(df_etf.index)

df_alinhado.head()
```

```python
print(df_alinhado.isna().sum())
```

```python
df_limpo = df_alinhado.ffill().bfill()
```

```python
print(df_limpo.isna().sum())
```

```python
df_limpo.to_csv('data/processed/cesi_limpo.csv')
```
